```python
import polars as pl
```

```python
df = (
    pl.read_parquet("../output.parquet")
    # Parse JSON string to struct
    .select([
    "commit_cid",
    "repo",
    "rev",
    "seq",
    "prev_cid",
    "time",
    "blob_cids",
    pl.col("ops").str.json_decode(
        pl.List(pl.Struct([
            pl.Field("action", pl.Utf8),
            pl.Field("cid", pl.Utf8),  # Now it's just a string
            pl.Field("path", pl.Utf8),
            pl.Field("record_type", pl.Utf8),
        ]))).alias("ops_parsed")
    ]).explode("ops_parsed").unnest("ops_parsed")
)
```
