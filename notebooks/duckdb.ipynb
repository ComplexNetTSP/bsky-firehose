```python
import duckdb
import json
df = duckdb.query("""
SELECT repo, 
    block_type, 
    -- Extract specific fields from JSON
    json(raw_json)->'$."$type"' as type,
    json(raw_json)->'$.createdAt' as created_at,
    json(raw_json)->'$.text' as text,
    json(raw_json)->'$.langs' as langs,
    -- Extract nested embed
    json(raw_json)->'$.embed."$type"' as embed_type,
    json(raw_json)->'$.embed.record.cid' as embed_cid,
    json(raw_json)->'$.embed.record.uri' as embed_uri,
    json(raw_json) AS full_data,
FROM '../output/blocks/year=2026/month=6/day=8/*.parquet'
WHERE block_type = 'app.bsky.feed.post'
""").df()
```
